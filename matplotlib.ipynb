```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

#2D line plot

```python
hours=[2,3,4,7,6,5,1]
days=[1,2,3,4,5,6,7]
plt.title("Student Study Trends during the week")
plt.xlabel("Days")
plt.ylabel("Hours")
plt.plot(days,hours)
```

```python
hours=[2,3,4,7,6,5,1]
hours_2=[1,3,2,4,5,6,1]
days=[1,2,3,4,5,6,7]
plt.title("Student Study Trends during the week")
plt.xlabel("Days")
plt.ylabel("Hours")
plt.plot(days,hours)
plt.plot(days,hours_2)
```

# 2D line plot form pandas Data

```python
df=pd.read_csv("enrollment_data.csv")
df
```

```python
plt.title("Students Enrollment Over the 8 Years")
plt.xlabel("Years")
plt.ylabel("Students Count")
plt.plot(df["Year"],df["Programming"],label='Programming')
plt.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing')
plt.legend()
```

# Legend loction

```python
plt.title("Students Enrollment Over the 8 Years")
plt.xlabel("Years")
plt.ylabel("Students Count")
plt.plot(df["Year"],df["Programming"],label='Programming')
plt.plot(df['Year'],df['Digital Marketing'],label='Digital Marketing')
plt.legend(loc='lower right')
```
