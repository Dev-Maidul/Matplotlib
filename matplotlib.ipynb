# Scatter Plot

```python
students=pd.read_csv('student_IQdata.csv')
students
```

```python
#Relation between Study Hours and IQ Score - postive correlation
plt.xlabel("Study Hours")
plt.ylabel("IQ Score")
plt.title("Relation between Study Hours and IQ Score")

plt.scatter(students['Study_Hour'], students['IQ_Score'], label="Study Hours vs IQ",color='red')

plt.grid()
plt.legend()
plt.show()
```

```python
#Relation between Shoe size and IQ Score - > no correlation
plt.xlabel("Shoe_Size")
plt.ylabel("IQ_Score")
plt.title("Relation between Shoe size and IQ Score - > no correlation")
plt.scatter(students['Shoe_Size'],students['IQ_Score'],label="Shoe Size vs IQ_Score")

plt.grid()
plt.show()
```

```python
#Relation between Chilling hour and IQ Score - > negative correlation
plt.xlabel("Chilling Hours")
plt.ylabel("IQ Score")
plt.title("Relation between Chilling hour and IQ Score - > negative correlation")
plt.scatter(students['Chilling_Hours'],students['IQ_Score'],color='green',label="Chilling Hours VS IQ Score")

plt.grid()
plt.legend()
```

# Histogram

```python
#based on iq 
plt.xlabel("IQ Score")
plt.ylabel("Frequency of Students")
plt.title("Frequency of Students based On IQ Score")
plt.hist(students['IQ_Score'],color='gray',edgecolor='red')
plt.grid()
```

# Bar Chart

```python
data = {
    'Student': [ 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender': [ 'Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male'],
    'StudyHours': [ 2, 5, 3, 7, 4, 8, 1],
    'Attendance': [ 60, 88, 70, 95, 80, 98, 55],
    'Grade': [ 'Fail', 'Pass', 'Fail', 'Pass', 'Pass', 'Pass', 'Fail']
}
df=pd.DataFrame(data)
df
```

```python
# gender wise 
gender=df.groupby('Gender').size()
plt.xlabel("Gender")
plt.ylabel("Count")
plt.title("Bar Chart of Gender")

plt.bar(gender.index,gender.values,color=['pink','black'],edgecolor='yellow')
plt.show()
```

```python
#  Bar chart on real data 
df=pd.read_csv('student_data.csv')
status=df.groupby('CompletionStatus').size()
plt.xlabel("Course Completion Status")
plt.ylabel("Count")
plt.title("Bar chart of Course Completion Status")
plt.bar(status.index,status.values,color=['green','yellow','red'],edgecolor='black')

# plt.grid()
plt.show()
```

```python
# location 

location=df.groupby('Location').size()
plt.xlabel("Location")
plt.ylabel("Counts")
plt.title("Bar Chart of Location")
plt.bar(location.index,location.values,edgecolor='red')
plt.show()
```

```python
instructors=df.groupby('Instructor').size()
plt.xlabel('Instructor')
plt.ylabel("Counts")
plt.title("Bar chart of Instructor")
plt.bar(instructors.index,instructors.values,edgecolor='pink',color=['black','blue','yellow'])
plt.show()
```

# Pie Chart

```python
status = df.groupby('CompletionStatus').size() 
plt.pie(status,labels=status.index, autopct='%1.1f%%',shadow=True, startangle=90)
plt.title("Pie Chart of Course Completetion Status")
plt.show()
```
