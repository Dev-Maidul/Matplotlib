#2D line plot

```python
hours=[2,3,4,7,6,5,1]
days=[1,2,3,4,5,6,7]
plt.title("Student Study Trends during the week")
plt.xlabel("Days")
plt.ylabel("Hours")
plt.plot(days,hours)
```
